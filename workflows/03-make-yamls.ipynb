# Make YAML deficit tracer configurations

```python
import yaml
```

```python
basins = [
    "North_Atlantic_basin",
    "North_Pacific_basin",
    "South",
    "Southern_Ocean",
]

npolygon = {
    "North_Atlantic_basin": 150,
    "North_Pacific_basin": 200,
    "South": 300,
    "Southern_Ocean": 40,
}
```

```python
# --- Build forward and reverse maps in one pass ---
polygon_master_map = {}       # (basin, polygon) → index
inverse_polygon_map = {}      # index → (basin, polygon)

idx = 0
for b in basins:
    for p in range(npolygon[b]):
        polygon_master_map[(b, p)] = idx
        inverse_polygon_map[idx] = (b, p)
        idx += 1
```

```python
cdr_forcing_root = "/global/cfs/projectdirs/m4746/Projects/OAE-Efficiency-Map/data/alk-forcing/OAE-Efficiency-Map"

def alk_forcing_file(b, p, d):
    return f"{cdr_forcing_root}/alk-forcing-{b}.{p:03d}-{d}.nc"
```

## All DOR, daily $\beta, \eta$

```python
date = "1999-01"
suffix = "all-dor-daily"

common_exp = {
    "varname": "alk_forcing",
    "is_alk": False,
    "year_first": 1999,
    "year_last": 2019,
    "year_align": 347,
    "tintalgo": "linear",
    "taxMode": "cycle",
}   

experiments = [
    {
        **common_exp,
        "basin": basin,
        "polygon": polygon,
        "forcing_file": alk_forcing_file(basin, polygon, date),
    }
    for basin in basins
    for polygon in range(npolygon[basin])
]

antitracer_config = {
    "suffix": suffix,
    "date": date,
    "experiments": experiments,
    "beta_forcing": {
        "file": "/global/cfs/projectdirs/m4746/Projects/OAE-Efficiency-Map/data/carbonate-sensitivities/carbonate_sensitivity_daily.nc",
        "varname": "dDICdCO2",
        "year_first": 346,
        "year_last": 368,
        "year_align": 346,
        "tintalgo": "linear",
        "taxMode": "extend",
    },
    "eta_forcing": {
        "file": "/global/cfs/projectdirs/m4746/Projects/OAE-Efficiency-Map/data/carbonate-sensitivities/carbonate_sensitivity_daily.nc",
        "varname": "dDICdALK",
        "year_first": 346,
        "year_last": 368,
        "year_align": 346,
        "tintalgo": "linear",
        "taxMode": "extend",
    },
}
```

```python
with open(f"antitracer-configs/antitracer_config_{date}_{suffix}.yaml", "w") as f:
    yaml.dump(antitracer_config, f)
```

## All OAE, daily $\beta, \eta$

```python
date = "1999-01"
suffix = "all-oae-daily"
common_exp = {
    "varname": "alk_forcing",
    "year_first": 1999,
    "year_last": 2019,
    "year_align": 347,
    "tintalgo": "linear",
    "taxMode": "cycle",
}
experiments = []
for basin in basins:
    for polygon in range(npolygon[basin]):
        # ALK tracer: receives alkalinity forcing directly, no gas exchange
        experiments.append({
            **common_exp,
            "basin": basin,
            "polygon": polygon,
            "forcing_file": alk_forcing_file(basin, polygon, date),
            "is_alk": True,
        })
        # DIC tracer: no direct source, gas exchange coupled to ALK partner above
        experiments.append({
            **common_exp,
            "basin": basin,
            "polygon": polygon,
            "forcing_file": None,
            "varname": None,
            "is_alk": False,
        })
antitracer_config = {
    "suffix": suffix,
    "date": date,
    "experiments": experiments,
    "beta_forcing": {
        "file": "/global/cfs/projectdirs/m4746/Projects/OAE-Efficiency-Map/data/carbonate-sensitivities/carbonate_sensitivity_daily.nc",
        "varname": "dDICdCO2",
        "year_first": 346,
        "year_last": 368,
        "year_align": 346,
        "tintalgo": "linear",
        "taxMode": "extend",
    },
    "eta_forcing": {
        "file": "/global/cfs/projectdirs/m4746/Projects/OAE-Efficiency-Map/data/carbonate-sensitivities/carbonate_sensitivity_daily.nc",
        "varname": "dDICdALK",
        "year_first": 346,
        "year_last": 368,
        "year_align": 346,
        "tintalgo": "linear",
        "taxMode": "extend",
    },
}
```

```python
with open(f"antitracer-configs/antitracer_config_{date}_{suffix}.yaml", "w") as f:
    yaml.dump(antitracer_config, f)
```

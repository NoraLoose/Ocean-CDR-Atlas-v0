# Make YAML deficit tracer configurations

```python
import yaml
```

## All

```python
suffix = "all"
date = "1999-01"
```

```python
basins = [
    "North_Atlantic_basin",
    "North_Pacific_basin",
    "South",
    "Southern_Ocean",
]

npolygon = {
    "North_Atlantic_basin": 150,
    "North_Pacific_basin": 200,
    "South": 300,
    "Southern_Ocean": 40,
}

antitracer_config = {
    "suffix": suffix,
    "date": date,
    "experiments": []
}

for basin in basins:
    for p in range(npolygon[basin]):
        antitracer_config["experiments"].append({"basin": basin, "polygon": p})

with open(f"antitracer-configs/antitracer_config_{date}_{suffix}.yaml", "w") as f:
    yaml.dump(antitracer_config, f)
```
